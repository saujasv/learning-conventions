```python
import json
import pandas as pd
import numpy as np
import random
from dotenv import load_dotenv
import os
from pathlib import Path

load_dotenv()
```

```python
lexgram_path = Path(os.getenv("LEXGRAM_PATH"))
kilogram_path = Path(os.getenv("KILOGRAM_PATH"))
```

```python
lexgram_df = pd.read_csv(lexgram_path / "tangram_data_semantic_uncertainty_paper" / "exp1_data.csv")
```

```python
lexgram_targets = [t.replace(".svg", "") for t in lexgram_df.target.unique()]
```

```python
with open(kilogram_path / "dataset" / "full.json") as f:
    kilogram_full_data = json.load(f)

with open(kilogram_path / "dataset" / "dense.json") as f:
    kilogram_dense_data = json.load(f)
```

```python
test_targets = set(kilogram_dense_data.keys()) | set(lexgram_targets)
training_targets = sorted(set(kilogram_full_data.keys()) - test_targets)
```

```python
random.Random(412).shuffle(training_targets)
pretraining_pool = training_targets[:300]
validation_pool = training_targets[300:350]
finetuning_pool = training_targets[350:]
```

```python
with open(kilogram_path / "dataset" / "pretraining_pool.json", "w") as f:
    json.dump({k: kilogram_full_data[k] for k in pretraining_pool}, f)
```

```python
with open(kilogram_path / "dataset" / "validation_pool.json", "w") as f:
    json.dump({k: kilogram_full_data[k] for k in validation_pool}, f)
```

```python
with open(kilogram_path / "dataset" / "finetuning_pool.json", "w") as f:
    json.dump({k: kilogram_full_data[k] for k in finetuning_pool}, f)
```

```python
with open(kilogram_path / "dataset" / "test_pool.json", "w") as f:
    json.dump({k: kilogram_full_data[k] for k in test_targets}, f)
```
